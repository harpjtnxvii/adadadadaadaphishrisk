```python
import os
import getpass
from pathlib import Path
from dotenv import load_dotenv

lokasi_env = direktori_project / ".env"
lokasi_gitignore = direktori_project / ".gitignore"

load_dotenv(lokasi_env)

def update_env_file(lokasi_file: Path, pasangan: dict[str, str]) -> None:
    isi_lama = lokasi_file.read_text(encoding="utf-8") if lokasi_file.exists() else ""
    baris = isi_lama.splitlines()
    hasil = []
    kunci_sudah = set()

    for item in baris:
        item_strip = item.strip()
        if not item_strip or item_strip.startswith("#") or "=" not in item:
            hasil.append(item)
            continue

        nama = item.split("=", 1)[0].strip()
        if nama in pasangan:
            hasil.append(f"{nama}={pasangan[nama]}")
            kunci_sudah.add(nama)
        else:
            hasil.append(item)

    for nama, nilai in pasangan.items():
        if nama not in kunci_sudah:
            hasil.append(f"{nama}={nilai}")

    lokasi_file.write_text("\n".join(hasil).strip() + "\n", encoding="utf-8")


def pastikan_gitignore_aman(lokasi_file: Path) -> None:
    isi = lokasi_file.read_text(encoding="utf-8") if lokasi_file.exists() else ""
    tambahan = [
        ".env",
        ".env.*",
        "!.env.example",
    ]

    baris = isi.splitlines()
    for item in tambahan:
        if item not in baris:
            baris.append(item)

    lokasi_file.write_text("\n".join(baris).strip() + "\n", encoding="utf-8")


api_key = os.getenv("OPENAI_API_KEY", "").strip()

if not api_key:
    api_key = getpass.getpass("Masukkan OPENAI_API_KEY lengkap kamu: ").strip()

if not api_key:
    raise ValueError("OPENAI_API_KEY masih kosong. Masukkan API key lengkap agar AI eksternal aktif.")

if not api_key.startswith("sk-"):
    print("Peringatan: format API key tidak diawali sk-. Tetap disimpan, tetapi cek lagi jika koneksi gagal.")

model_default = os.getenv("OPENAI_MODEL", "gpt-5.5-mini").strip() or "gpt-5.5-mini"

update_env_file(
    lokasi_env,
    {
        "OPENAI_API_KEY": api_key,
        "OPENAI_MODEL": model_default,
        "PHISHRISK_AI_MODE": "openai",
    }
)

pastikan_gitignore_aman(lokasi_gitignore)

os.environ["OPENAI_API_KEY"] = api_key
os.environ["OPENAI_MODEL"] = model_default
os.environ["PHISHRISK_AI_MODE"] = "openai"

print("API key berhasil disiapkan untuk runtime lokal.")
print("File .env dibuat/diperbarui di:", lokasi_env)
print("Model AI:", model_default)
print("Gitignore sudah diamankan agar .env tidak ikut GitHub.")
print("API key tidak ditampilkan.")
```

```python
# Tes koneksi OpenAI API versi aman.
# Jika API gagal, notebook tetap lanjut memakai fallback lokal.

import os
import pandas as pd
from openai import OpenAI

api_key = os.getenv("OPENAI_API_KEY", "").strip()

if not api_key:
    raise ValueError("OPENAI_API_KEY belum terbaca. Jalankan ulang cell input API key terlebih dahulu.")

client = OpenAI(api_key=api_key)

model_kandidat = []

model_env = os.getenv("OPENAI_MODEL", "").strip()
if model_env:
    model_kandidat.append(model_env)

# Urutan model dibuat lebih aman:
# 1. Model ringan yang biasanya lebih murah
# 2. Model yang lebih kuat jika tersedia
# 3. Model lama sebagai cadangan
for nama_model in [
    "gpt-4.1-mini",
    "gpt-4o-mini",
    "gpt-5.5",
    "gpt-5.4-mini",
]:
    if nama_model not in model_kandidat:
        model_kandidat.append(nama_model)

hasil_tes_model = []
model_berhasil = None
teks_berhasil = ""

for nama_model in model_kandidat:
    try:
        response = client.responses.create(
            model=nama_model,
            input=(
                "Jawab singkat dalam bahasa Indonesia. "
                "Tulis satu kalimat bahwa AI PhishRisk aktif untuk penjelasan defensif."
            ),
        )

        teks_output = getattr(response, "output_text", "").strip()

        hasil_tes_model.append({
            "model": nama_model,
            "status": "berhasil",
            "pesan": teks_output[:300],
        })

        model_berhasil = nama_model
        teks_berhasil = teks_output
        break

    except Exception as error:
        pesan_error = str(error)

        if "429" in pesan_error:
            jenis_error = "quota_atau_rate_limit"
        elif "400" in pesan_error:
            jenis_error = "model_tidak_tersedia_atau_request_tidak_valid"
        elif "401" in pesan_error:
            jenis_error = "api_key_tidak_valid"
        else:
            jenis_error = "error_lain"

        hasil_tes_model.append({
            "model": nama_model,
            "status": "gagal",
            "jenis_error": jenis_error,
            "pesan": pesan_error[:500],
        })

data_tes_model = pd.DataFrame(hasil_tes_model)
display(data_tes_model)

if model_berhasil:
    os.environ["OPENAI_MODEL"] = model_berhasil
    os.environ["PHISHRISK_AI_EXTERNAL_ACTIVE"] = "1"

    update_env_file(lokasi_env, {
        "OPENAI_MODEL": model_berhasil,
        "PHISHRISK_AI_EXTERNAL_ACTIVE": "1",
    })

    print("AI eksternal aktif.")
    print("Model berhasil:", model_berhasil)
    print("Output tes:", teks_berhasil)

else:
    os.environ["PHISHRISK_AI_EXTERNAL_ACTIVE"] = "0"

    update_env_file(lokasi_env, {
        "PHISHRISK_AI_EXTERNAL_ACTIVE": "0",
    })

    print("AI eksternal belum aktif.")
    print("Program tetap lanjut memakai fallback lokal.")
    print("Kemungkinan penyebab:")
    print("- API quota/billing belum aktif atau habis.")
    print("- Model tidak tersedia untuk akun kamu.")
    print("- Usage limit akun API perlu dinaikkan.")
    print("- API key benar, tapi belum punya akses penggunaan.")
```

```python
from pathlib import Path

blok_dotenv = (
    "# Auto-load environment lokal PhishRisk.\n"
    "try:\n"
    "    from pathlib import Path as _PhishRiskPath\n"
    "    from dotenv import load_dotenv as _phishrisk_load_dotenv\n"
    "    _phishrisk_load_dotenv(_PhishRiskPath(__file__).resolve().parents[1] / \".env\")\n"
    "except Exception:\n"
    "    pass\n"
)

def sisipkan_blok_dotenv(lokasi_file: Path) -> None:
    isi = lokasi_file.read_text(encoding="utf-8")

    if "Auto-load environment lokal PhishRisk" in isi:
        print(f"Sudah ada auto-load .env: {lokasi_file.name}")
        return

    lines = isi.splitlines()
    index_sisip = 0

    for i, line in enumerate(lines):
        if line.startswith("from __future__"):
            index_sisip = i + 1

    lines.insert(index_sisip, blok_dotenv.strip())
    lokasi_file.write_text("\n".join(lines).strip() + "\n", encoding="utf-8")
    print(f"Auto-load .env ditambahkan ke: {lokasi_file.name}")

for nama_file in ["ai_explainer.py", "run_ai_phishrisk.py"]:
    lokasi = direktori_src / nama_file
    sisipkan_blok_dotenv(lokasi)

print("Patch auto-load .env selesai.")
```
