## Import Library

```python
from pathlib import Path
import pandas as pd
import numpy as np
```

## Mengatur Lokasi File

```python
nama_file_dataset = "PhiUSIIL_Phishing_URL_Dataset.csv"

direktori_project = Path.cwd()
lokasi_dataset_utama = direktori_project / nama_file_dataset
lokasi_dataset_alternatif = direktori_project / "data" / "raw" / nama_file_dataset

direktori_laporan = direktori_project / "reports" / "outputs"
direktori_laporan.mkdir(parents=True, exist_ok=True)

if lokasi_dataset_utama.exists():
    lokasi_dataset = lokasi_dataset_utama
elif lokasi_dataset_alternatif.exists():
    lokasi_dataset = lokasi_dataset_alternatif
else:
    raise FileNotFoundError("Dataset tidak ditemukan. Pastikan file CSV berada di folder utama project atau di data/raw.")

print("Lokasi project:", direktori_project)
print("Lokasi dataset:", lokasi_dataset)
```

## Membaca Dataset

```python
data_phishing = pd.read_csv(lokasi_dataset)

print("Dataset berhasil dibaca.")
print("Jumlah baris:", data_phishing.shape[0])
print("Jumlah kolom:", data_phishing.shape[1])
```

## Menampilkan Data Awal

```python
data_phishing.head()
```

## Menampilkan Data Akhir

```python
data_phishing.tail()
```

## Mengecek Nama Kolom

```python
daftar_kolom = pd.DataFrame({
    "nomor": range(1, len(data_phishing.columns) + 1),
    "nama_kolom": data_phishing.columns
})

daftar_kolom
```

## Mengecek Informasi Tipe Data

```python
ringkasan_tipe_data = pd.DataFrame({
    "nama_kolom": data_phishing.columns,
    "tipe_data": data_phishing.dtypes.astype(str),
    "jumlah_data_terisi": data_phishing.notna().sum().values,
    "jumlah_data_kosong": data_phishing.isna().sum().values,
    "jumlah_nilai_unik": data_phishing.nunique().values
})

ringkasan_tipe_data
```

## Mengecek Missing Value

```python
ringkasan_missing_value = pd.DataFrame({
    "nama_kolom": data_phishing.columns,
    "jumlah_missing": data_phishing.isna().sum().values,
    "persentase_missing": (data_phishing.isna().mean().values * 100).round(2)
})

ringkasan_missing_value = ringkasan_missing_value.sort_values(
    by="jumlah_missing",
    ascending=False
)

ringkasan_missing_value
```

## Mengecek Duplikasi Data

```python
jumlah_duplikasi = data_phishing.duplicated().sum()
persentase_duplikasi = round((jumlah_duplikasi / len(data_phishing)) * 100, 2)

print("Jumlah data duplikat:", jumlah_duplikasi)
print("Persentase data duplikat:", persentase_duplikasi, "%")
```

## Mengecek Kolom Target

```python
nama_kolom_target = "label"

if nama_kolom_target not in data_phishing.columns:
    raise ValueError("Kolom target 'label' tidak ditemukan. Periksa kembali nama kolom target pada dataset.")

print("Kolom target ditemukan:", nama_kolom_target)
print("Nilai unik pada target:", sorted(data_phishing[nama_kolom_target].unique()))
```

## Mengecek Distribusi Label

```python
distribusi_label = data_phishing[nama_kolom_target].value_counts().sort_index()
persentase_label = data_phishing[nama_kolom_target].value_counts(normalize=True).sort_index() * 100

ringkasan_label = pd.DataFrame({
    "label": distribusi_label.index,
    "jumlah_data": distribusi_label.values,
    "persentase": persentase_label.round(2).values
})

ringkasan_label["keterangan"] = ringkasan_label["label"].map({
    0: "Phishing",
    1: "Legitimate"
})

ringkasan_label
```

## Mengecek Kolom Numerik dan Kategorikal

```python
kolom_numerik = data_phishing.select_dtypes(include=["int64", "float64"]).columns.tolist()
kolom_kategorikal = data_phishing.select_dtypes(include=["object"]).columns.tolist()

print("Jumlah kolom numerik:", len(kolom_numerik))
print("Jumlah kolom kategorikal:", len(kolom_kategorikal))

print("\nKolom kategorikal:")
print(kolom_kategorikal)
```

## Statistik Deskriptif Kolom Numerik

```python
data_phishing[kolom_numerik].describe().T
```

## Mengecek Kolom dengan Nilai Unik Terlalu Banyak

```python
ringkasan_nilai_unik = pd.DataFrame({
    "nama_kolom": data_phishing.columns,
    "jumlah_nilai_unik": data_phishing.nunique().values,
    "persentase_nilai_unik": ((data_phishing.nunique().values / len(data_phishing)) * 100).round(2)
})

ringkasan_nilai_unik = ringkasan_nilai_unik.sort_values(
    by="jumlah_nilai_unik",
    ascending=False
)

ringkasan_nilai_unik.head(15)
```

## Menyimpan Ringkasan Validasi

```python
lokasi_ringkasan_tipe_data = direktori_laporan / "ringkasan_tipe_data.csv"
lokasi_ringkasan_missing_value = direktori_laporan / "ringkasan_missing_value.csv"
lokasi_ringkasan_label = direktori_laporan / "ringkasan_label.csv"
lokasi_ringkasan_nilai_unik = direktori_laporan / "ringkasan_nilai_unik.csv"

ringkasan_tipe_data.to_csv(lokasi_ringkasan_tipe_data, index=False)
ringkasan_missing_value.to_csv(lokasi_ringkasan_missing_value, index=False)
ringkasan_label.to_csv(lokasi_ringkasan_label, index=False)
ringkasan_nilai_unik.to_csv(lokasi_ringkasan_nilai_unik, index=False)

print("Ringkasan validasi berhasil disimpan.")
print("Folder output:", direktori_laporan)
```

## Kesimpulan Validasi Awal

```python
jumlah_baris, jumlah_kolom = data_phishing.shape
jumlah_missing_total = data_phishing.isna().sum().sum()
jumlah_kolom_numerik = len(kolom_numerik)
jumlah_kolom_kategorikal = len(kolom_kategorikal)

print("KESIMPULAN VALIDASI AWAL")
print("-" * 40)
print("Jumlah baris:", jumlah_baris)
print("Jumlah kolom:", jumlah_kolom)
print("Jumlah kolom numerik:", jumlah_kolom_numerik)
print("Jumlah kolom kategorikal:", jumlah_kolom_kategorikal)
print("Total missing value:", jumlah_missing_total)
print("Jumlah data duplikat:", jumlah_duplikasi)
print("Kolom target:", nama_kolom_target)
print("\nDistribusi label:")
print(ringkasan_label)
```
